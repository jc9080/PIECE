```python
from pathlib import Path
import pandas as pd
import numpy as np

# Define project paths
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "raw"

# Load original datasets
df_vmr = pd.read_csv(DATA_DIR / "vmr_all.csv")
df_raw = pd.read_csv(DATA_DIR / "ige_ege_jump.csv")

# Check dataset dimensions
print("VMR dataset:", df_vmr.shape)
print("Raw dataset:", df_raw.shape)
```

```python
# Inspect column names
print("VMR columns:")
print(df_vmr.columns.tolist())

print("\nRaw data columns:")
print(df_raw.columns.tolist())

# Preview dataset
display(df_vmr.head())
```

```python
# 1. Number of participants
print("Participants:", df_vmr["SN"].nunique())

# 2. Number of trials per participant
print("\nTrials per participant:")
print(df_vmr.groupby("SN").size())

# 3. Rotation conditions
print("\nRotation conditions:")
print(df_vmr["rotation"].value_counts(dropna=False).sort_index())

# 4. Missing values
print("\nMissing values:")
print(df_vmr.isnull().sum().loc[lambda x: x > 0])
```

```python
# Dataset validation
print("DATASET VALIDATION")
print("=" * 45)

print("\n1. Rotation distribution")
print(
    df_vmr["rotation"]
    .value_counts(dropna=False)
    .sort_index()
    .to_string()
)

print("\n2. Missing values in key variables")

key_vars = [
    "rotation",
    "theta_maxradv",
    "theta_maxradv_clean",
    "adaptation",
    "motor_sd",
    "perturbation"
]

print(
    df_vmr[key_vars]
    .isna()
    .sum()
    .to_string()
)

print("\n3. Perturbation conditions")
print(
    df_vmr["perturbation"]
    .value_counts(dropna=False)
    .to_string()
)
```
